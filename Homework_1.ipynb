#**Q3. Fuel types**


```python
#Check unique fuel types will simply tell how many fuel types presented in the dataset

#Show all unique fuel types
different_fuels = df['fuel_type'].unique()
print("Different unique fuel types:", different_fuels)

#Count how many unique fuel types there are
num_different_fuels = df['fuel_type'].nunique()
print("Number of different fuel types:", num_different_fuels)

# I can also see the distribution of each fuel type
display(df['fuel_type'].value_counts())


```

#**Q4. Missing values**

```python
#Check how many missing values are in each column
missing_per_column = df.isnull().sum()
print("Missing values per column:")
print(missing_per_column)

# Check how many columns have at least one missing value
columns_with_missing = missing_per_column[missing_per_column > 0]
num_columns_with_missing = len(columns_with_missing)

print(f"\nNumber of columns with missing values: {num_columns_with_missing}")
if num_columns_with_missing > 0:
    print("Columns with missing values:", list(columns_with_missing.index))
```

#**Q5. Max fuel efficiency**

```python

asia_cars = df[df['origin'] == 'Asia']
max_asia_fuel_efficiency = asia_cars['fuel_efficiency_mpg'].max()
print(f"The maximum fuel efficiency of cars from Asia is: {max_asia_fuel_efficiency}")
```

#**Q6. Median value of horsepower**


```python
# I calculated the initial median value of the horsepower column
initial_median = df['horsepower'].median()
print(f"Initial median of horsepower: {initial_median}")

# Find the most frequent value (mode) of the horsepower column
# I used [0] to get the actual value from .mode()
most_frequent_horsepower = df['horsepower'].mode()[0]
print(f"Most frequent horsepower value: {most_frequent_horsepower}")

# Fill the missing values in 'horsepower' with the most frequent value
df['horsepower'] = df['horsepower'].fillna(most_frequent_horsepower)

# Calculate the median value of horsepower once again
new_median = df['horsepower'].median()
print(f"New median of horsepower (after filling missing values): {new_median}")
```


#**Q7. Sum of weights**


```python
import numpy as np

# Select all the cars from Asia
asia_cars = df[df['origin'] == 'Asia']

# Select only columns 'vehicle_weight' and 'model_year'
subset_df = asia_cars[['vehicle_weight', 'model_year']]

# Select the first 7 values
first7_df = subset_df.head(7)
display(first7_df)

# Get the underlying NumPy array (X)
X = first7_df.values
print(X)

# Compute matrixmatrix multiplication between transpose of X (X.T) and X

XTX = X.T @ X

print(XTX)

# Invert XTX
XTX_inv = np.linalg.inv(XTX)

print(XTX_inv)
```

```python
# Create an array y with the specified values
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# Multiply the inverse of XTX with the transpose of X, and then multiply by y
w = (XTX_inv @ X.T) @ y

print(w)

# The sum of all the elements of the result w

sum_w = w.sum()
print(f"Sum of all elements of w: {sum_w:}")
```
